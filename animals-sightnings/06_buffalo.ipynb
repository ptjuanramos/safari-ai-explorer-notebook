## Background (pseudo-absence) points

A presence-only model needs something to compare presences against. Background points are a sample of "what the landscape looks like in general", not confirmed absences.

**The catch:** GBIF sightings come from tourists, so they pile up around roads and lodges. Background spread over the whole box would let the model learn "near roads vs remote" instead of buffalo habitat. To avoid that:

1. **Visited area:** download every mammal sighting (any species) in the box. A 10 km grid cell counts as *visited* if it has at least one. Background is only drawn inside visited cells, so both classes come from places people actually go.
2. **Thinning:** keep one buffalo sighting per 1 km cell per year, so a popular spot reported 40 times counts once.

How `make_presence_background` then builds the background:
1. **Seed:** one `np.random.default_rng(seed)` generator, so every run gives the same points.
2. **Meters, not degrees (ADR-0001):** points are drawn uniformly in EPSG:32736 (UTM 36S) and converted back to lon/lat. Points outside the lon/lat box or outside a visited cell are thrown away.
3. **Years:** each presence year is repeated 10 times (`np.repeat`), so the background year distribution matches the presences exactly. This matters because embeddings change year to year.
4. **No near-duplicates:** the embeddings average a `BUFFER_IN_METERS` buffer, so a background point closer than that to a same-year presence would look almost identical to it. Such points are redrawn.
5. **Label:** `presence = 1` for sightings and `0` for background. Both go into one DataFrame.

## Embeddings for all points

That gives `sdm_df`, the table the model will use.

## Validation

Four checks before modelling: duplicates, year/class balance, spatial coverage and separability. Distances are computed in meters (EPSG:32736). scikit-learn (which brings scipy) is needed for the KD-tree and the quick model.

```python
%pip install scikit-learn
```

```python
from scipy.spatial import cKDTree

geo = sdm_df[["year", "presence"]].copy()
geo["x"], geo["y"] = to_utm.transform(sdm_df["decimalLongitude"].values, sdm_df["decimalLatitude"].values)

pres = geo[geo["presence"] == 1]
bg = geo[geo["presence"] == 0]
```

### Check 1: no background point duplicates a presence

For each year, a KD-tree (a fast nearest-neighbour index) is built on that year's presences. It gives the distance from every background point to its closest same-year presence, and the smallest of those distances must be at least `BUFFER_IN_METERS`. Exact lon/lat matches are also counted. The expected count is 0.

```python
bg_min_dist = pd.Series(index=bg.index, dtype=float)
for year, bg_year in bg.groupby("year"):
    tree = cKDTree(pres.loc[pres["year"] == year, ["x", "y"]])
    bg_min_dist[bg_year.index] = tree.query(bg_year[["x", "y"]])[0]

exact_matches = bg_rows.merge(pres_rows, on=["decimalLongitude", "decimalLatitude"])

print(f"closest background point to a same-year presence: {bg_min_dist.min():.0f} m")
print("exact coordinate matches:", len(exact_matches))
assert bg_min_dist.min() >= BUFFER_IN_METERS, f"a background point is within {BUFFER_IN_METERS} m of a same-year presence"
```

### Check 2: year and class balance

- Expected counts: the thinned presence count, and 10× that for background.
- Every year should have exactly 10 background points per presence.
- The two classes differ **only in the label** by construction: they share one table and were processed by one Earth Engine call. Missing embeddings per class were already printed in the extraction cell above.

```python
print("class counts:")
print(sdm_df["presence"].value_counts())

year_table = pd.crosstab(sdm_df["year"], sdm_df["presence"])
year_table["ratio"] = year_table[0] / year_table[1]
print("\nyear distribution per class:")
print(year_table)
assert (year_table["ratio"] == 10).all(), "background/presence ratio is not 10 in every year"

print("\nyear summary per class:")
print(sdm_df.groupby("presence")["year"].describe())
```

### Check 3: spatial coverage

Background points should spread evenly over the box. Sightings usually don't, because people report from roads, lodges and popular spots. Three numeric summaries:
- **Grid counts:** the box is cut into 10 km cells, and presences and background are counted per cell. If a small share of cells holds most of the presences, the sightings are concentrated.
- **Nearest-presence distance:** for each presence, the distance to the nearest *other* presence. For each background point, the distance to the nearest presence. If presences sit much closer to each other than background sits to them, they are clustered.
- **Clark–Evans ratio R:** the mean nearest-neighbour distance divided by what random placement would give. R ≈ 1 is random and R < 1 is clustered. Clustering is flagged as heavy if R < 0.5 or the top 10 % of cells hold over half the presences.

```python
from shapely.geometry import Polygon

geo["cell"] = (geo["x"] // cell_size_m).astype(int).astype(str) + "_" + (geo["y"] // cell_size_m).astype(int).astype(str)
cell_counts = pd.crosstab(geo["cell"], geo["presence"])

pres_per_cell = cell_counts[1].sort_values(ascending=False)
n_top = int(np.ceil(0.1 * len(cell_counts)))
top_share = pres_per_cell.iloc[:n_top].sum() / pres_per_cell.sum()

print("grid cells with background:", (cell_counts[0] > 0).sum())
print("grid cells with >=1 presence:", (cell_counts[1] > 0).sum())
print("max presences in one cell:", pres_per_cell.max())
print(f"share of presences in the top 10% of cells ({n_top} cells): {top_share:.0%}")

tree = cKDTree(pres[["x", "y"]])
bg_nn = tree.query(bg[["x", "y"]], k=1)[0]
pres_nn = tree.query(pres[["x", "y"]], k=2)[0][:, 1]

print("\ndistance to nearest presence (m):")
print(pd.DataFrame({
    "presence -> other presence": pd.Series(pres_nn).describe(),
    "background -> presence": pd.Series(bg_nn).describe(),
}).round(0))

min_lon, min_lat, max_lon, max_lat = serengeti_bounds
box_x, box_y = to_utm.transform([min_lon, max_lon, max_lon, min_lon], [min_lat, min_lat, max_lat, max_lat])
box_area = Polygon(zip(box_x, box_y)).area
expected_nn = 0.5 / np.sqrt(len(pres) / box_area)
clark_evans = pres_nn.mean() / expected_nn

heavy_clustering = clark_evans < 0.5 or top_share > 0.5
print(f"\nClark-Evans R: {clark_evans:.2f} (1 = random, <1 = clustered)")
print("heavy clustering of presences:", heavy_clustering)
```

### Check 4: can the embeddings tell the classes apart?

Fit one quick model (standardize, then logistic regression) on the 64 embedding bands only, and score it with cross-validated AUC. `class_weight="balanced"` stops the 10:1 imbalance from dominating.

The same model is scored two ways:
- **Random CV:** points are hidden at random. A hidden sighting usually has a near-twin a few hundred meters away in the training data, so this score is optimistic.
- **Block CV:** whole 10 km cells are hidden at once (`StratifiedGroupKFold`), so the model must predict areas it has never seen. This is the honest number.

How to read it:
- **Block AUC ≈ 0.5:** the embeddings don't distinguish buffalo locations from the visited landscape.
- **Block AUC well above 0.5:** real habitat signal.
- **Big gap between random and block:** that's how much the clustering was inflating the score.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedGroupKFold, StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X = sdm_df[band_cols]
y = sdm_df["presence"]

model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, class_weight="balanced"))

random_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
block_cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)

auc_random = cross_val_score(model, X, y, cv=random_cv, scoring="roc_auc")
auc_block = cross_val_score(model, X, y, cv=block_cv, groups=geo["cell"], scoring="roc_auc")

print(f"random CV AUC: {auc_random.mean():.3f} ± {auc_random.std():.3f}")
print(f"block CV AUC:  {auc_block.mean():.3f} ± {auc_block.std():.3f}")
print(f"inflation from clustering: {auc_random.mean() - auc_block.mean():+.3f}")
```

### Does a non-linear model find more signal?

Logistic regression can only use straight-line combinations of the 64 bands. Gradient boosting builds many small decision trees, each one fixing the previous ones' mistakes, so it can pick up curved effects and interactions, such as "band A matters only when band B is high". It doesn't need scaling.

Both models are scored with the **same block CV folds**, so the comparison is fair:
- **Clear gain (e.g. +0.03 or more):** the embeddings hold habitat signal the straight-line model couldn't use.
- **Little or no gain:** about 0.68 is roughly what these features can do, and logistic regression is good enough.

Per-fold scores are shown too. Folds that score low are regions where buffalo locations are harder to predict from the embeddings.

```python
from sklearn.ensemble import HistGradientBoostingClassifier

boosting = HistGradientBoostingClassifier(class_weight="balanced", random_state=42)
auc_boost = cross_val_score(boosting, X, y, cv=block_cv, groups=geo["cell"], scoring="roc_auc")

print("block CV AUC per fold:")
print(pd.DataFrame({"logistic": auc_block, "boosting": auc_boost}).round(3))
print(f"\nlogistic: {auc_block.mean():.3f} ± {auc_block.std():.3f}")
print(f"boosting: {auc_boost.mean():.3f} ± {auc_boost.std():.3f}")
print(f"gain from non-linear model: {auc_boost.mean() - auc_block.mean():+.3f}")
```

### XGBoost

XGBoost is another gradient-boosting library: the same idea of many small trees, each one fixing the previous ones' mistakes. It's widely used and has more tuning options for later. Two differences from `HistGradientBoostingClassifier`:
- It has no `class_weight`. Instead, `scale_pos_weight` = background count ÷ presence count (about 10) gives each presence about 10× the weight, which has the same effect.
- Its defaults differ (deeper trees, faster learning rate), so the scores won't be identical even though the method is similar.

It uses the **same block CV folds**, so all three models are compared on exactly the same held-out areas. Settings stay at their defaults. Tuning them against these folds would make the AUC optimistic again.

```python
%pip install xgboost
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    scale_pos_weight=(y == 0).sum() / (y == 1).sum(),
    eval_metric="auc",
    random_state=42,
    n_jobs=-1,
)
auc_xgb = cross_val_score(xgb, X, y, cv=block_cv, groups=geo["cell"], scoring="roc_auc")

print("block CV AUC per fold:")
print(pd.DataFrame({"logistic": auc_block, "hist boosting": auc_boost, "xgboost": auc_xgb}).round(3))
print(f"\nhist boosting: {auc_boost.mean():.3f} ± {auc_boost.std():.3f}")
print(f"xgboost:       {auc_xgb.mean():.3f} ± {auc_xgb.std():.3f}")
```

## Prediction map

Cross-validation told us how good the model is. Now train it once on **all** the data and ask it where buffalo should be.

1. **Final model:** the same XGBoost settings, fit on every presence and background point.
2. **Grid:** a regular grid of points in meters (`GRID_SPACING_M`, 2.5 km) inside every visited 10 km cell, for one year (`MAP_YEAR`). Unvisited cells are left out, because the model never saw background there and its predictions would be guesses.
3. **Embeddings:** each grid point goes through the **same** `get_range_year_reduced_area_embeddings` function, in chunks of 1,000 so each Earth Engine request stays small.
4. **Score:** `predict_proba` gives each point a score from 0 to 1.
5. **Cut-off for a cleaner map:** only points scoring at least `SUITABILITY_MIN` are drawn. The default is the score that 90 % of known buffalo sightings reach, a common rule for presence-only models. It's computed from out-of-fold block-CV scores: the final model has already seen every sighting and would score them all near 1, which would make the cut-off meaningless. You can overwrite it with your own number.

**Read the score as relative suitability, not probability.** `scale_pos_weight` and the 10:1 background ratio shift the numbers, so 0.8 doesn't mean an 80 % chance of buffalo. It means "looks more like buffalo habitat than a point scoring 0.4". Use it to rank and compare places.

> **Improved model:** this map uses the first model: all sightings pooled, embeddings only. Notebooks 07–10 found it transfers poorly to unseen regions, because half the sightings come from the Ngorongoro area, a different landscape. The recommended Serengeti map is in `10_buffalo_bias_checks.ipynb`: Serengeti only, regularized XGBoost, embeddings + habitat covariates.

```python
final_model = xgb.fit(X, y)
```

```python
MAP_YEAR = 2025
GRID_SPACING_M = 2_500
GEE_CHUNK = 1_000

offsets = np.arange(GRID_SPACING_M / 2, cell_size_m, GRID_SPACING_M)
grid = pd.DataFrame(
    [(cx * cell_size_m + ox, cy * cell_size_m + oy) for cx, cy in visited_cells for ox in offsets for oy in offsets],
    columns=["x", "y"],
)
grid["decimalLongitude"], grid["decimalLatitude"] = to_lonlat.transform(grid["x"].values, grid["y"].values)
grid = grid[grid["decimalLongitude"].between(min_lon, max_lon) & grid["decimalLatitude"].between(min_lat, max_lat)].reset_index(drop=True)
grid["point_id"] = "grid_" + grid.index.astype(str)
grid["year"] = MAP_YEAR
print("grid points:", len(grid))

chunks = []
for start in range(0, len(grid), GEE_CHUNK):
    info = get_range_year_reduced_area_embeddings(grid.iloc[start:start + GEE_CHUNK], "point_id", start_year=MAP_YEAR, end_year=MAP_YEAR)
    chunks.append(pd.DataFrame([f["properties"] for f in info[0]["features"]]))

grid = grid.merge(pd.concat(chunks, ignore_index=True), on="point_id", how="left")
print("missing embeddings:", grid[band_cols].isna().any(axis=1).sum())
grid = grid.dropna(subset=band_cols)

grid["suitability"] = final_model.predict_proba(grid[band_cols])[:, 1]
print(grid["suitability"].describe().round(3))
```

```python
import plotly.express as px
from sklearn.model_selection import cross_val_predict

oof = cross_val_predict(xgb, X, y, cv=block_cv, groups=geo["cell"], method="predict_proba")[:, 1]
SUITABILITY_MIN = np.quantile(oof[y.values == 1], 0.10)
SUITABILITY_MIN = 0.3

map_df = grid[grid["suitability"] >= SUITABILITY_MIN]
print(f"threshold: {SUITABILITY_MIN:.2f} -> keeping {len(map_df)} of {len(grid)} grid points ({len(map_df) / len(grid):.0%})")

BLUES = [[0, "#86b6ef"], [0.5, "#2a78d6"], [1, "#0d366b"]]

min_lon, min_lat, max_lon, max_lat = serengeti_bounds
fig = px.scatter_map(
    map_df,
    lat="decimalLatitude",
    lon="decimalLongitude",
    color="suitability",
    color_continuous_scale=BLUES,
    range_color=[SUITABILITY_MIN, map_df["suitability"].max()],
    hover_data={"suitability": ":.2f", "decimalLatitude": False, "decimalLongitude": False},
    center={"lat": (min_lat + max_lat) / 2, "lon": (min_lon + max_lon) / 2},
    zoom=7,
    height=650,
    map_style="carto-positron",
    title=f"Buffalo relative habitat suitability ≥ {SUITABILITY_MIN:.2f}, {MAP_YEAR} (XGBoost on AlphaEarth embeddings)",
)
fig.update_traces(marker_size=8)
fig.update_layout(coloraxis_colorbar=dict(title="suitability", thickness=12), margin=dict(t=60, l=0, r=0, b=0))
fig.show()
```

## Sanity check: Snapshot Serengeti camera traps

The camera traps are a completely **independent** source: fixed cameras that ran for a known number of days, not tourists with phones. That gives something GBIF never has: **effort**. A camera that ran 300 days and photographed buffalo 30 times tells us more than "someone once saw a buffalo here".

For each camera:
- **Detection rate** = confident buffalo captures (volunteer agreement ≥ 0.5, as in notebook 03) per 100 trap-days.
- **Predicted suitability** = the model's score at the camera's location, using the same embedding function.

If the model has learned something real, cameras in high-suitability spots should photograph buffalo more often:
- **Spearman rho:** a rank correlation from −1 to 1. Above 0 means higher scores go with higher detection rates, and it doesn't assume a straight-line relation.
- **Terciles:** cameras split into low / medium / high predicted suitability, with the mean detection rate of each group. It should rise from low to high.

**Read this as a sanity check, not a test score:**
- The cameras ran 2010–2013, but embeddings start in 2017, so 2017 is used as the closest year.
- The cameras cover only a small central area (about 1,100 km²), and 182 of 225 photographed buffalo at least once.
- Cameras that ran fewer than `CONFIG.min_trap_days` (30) days are dropped (3 sites), because one lucky photo would give them a huge rate.
- Detection rates also depend on things the model never sees: seasons, migration, and where each camera was pointed.

A weak positive relation is a good sign. A clearly negative one would mean something is wrong.

```python
from scipy.stats import spearmanr

MIN_AGREEMENT = CONFIG.min_agreement

cam = buffalo_data.load_camera_sites(CONFIG)
cam = buffalo_data.embed_points(cam, "SiteID", CONFIG).dropna(subset=band_cols)
cam["suitability"] = final_model.predict_proba(cam[band_cols])[:, 1]

rho, p_value = spearmanr(cam["suitability"], cam["rate"])
print(f"camera sites: {len(cam)}, with buffalo: {(cam['buffalo_captures'] > 0).sum()}")
print(f"Spearman rho (suitability vs detection rate): {rho:.2f} (p = {p_value:.3f})")

cam["suitability_group"] = pd.qcut(cam["suitability"], 3, labels=["low", "medium", "high"])
print("\nbuffalo captures per 100 trap-days, by predicted suitability:")
print(cam.groupby("suitability_group", observed=True)["rate"].agg(["count", "mean", "median"]).round(2))
```

## From score to probability: calibrating with the camera traps

The model's score is relative: 0.8 is "better than 0.4", not an 80 % chance. Presence-only data can't give real probabilities, because it never records where buffalo were *absent*. The camera traps do: each camera ran for a known number of days, and on most of them it photographed no buffalo.

1. **Detection days:** for each camera, count the days with at least one confident buffalo photo, out of its trap-days.
2. **Fit:** a logistic regression of "buffalo photographed that day: yes/no" on the model score. Each camera counts once per day it ran (`sample_weight`), and the score enters on the logit scale so 0.01 and 0.99 are not squashed together.
3. **Result:** a function from score to **daily detection probability**, the chance that a camera at that spot photographs a buffalo on a given day.

**What this probability is, and isn't:**
- It is per **camera view** (tens of meters), per **day**. It is not "the chance a buffalo lives in this 2.5 km cell".
- It was learned in the central camera area, from 2010–2013 detections and 2017 embeddings. Scores outside the range the cameras saw are **capped** to that range, so the probability never extrapolates past what the cameras observed.
- Areas are **not** combined into one "chance of buffalo in this area": buffalo move in herds, so neighbouring spots are not independent and that number would be inflated.

```python
def to_logit(score):
    score = np.clip(score, 1e-4, 1 - 1e-4)
    return np.log(score / (1 - score))


photos = pd.read_csv("data/consensus_data.csv", usecols=["SiteID", "DateTime", "Species", "NumVotes", "NumClassifications"])
photos = photos[(photos["Species"] == "buffalo") & (photos["NumVotes"] / photos["NumClassifications"] >= MIN_AGREEMENT)]
detection_days = photos.assign(day=photos["DateTime"].str[:10]).groupby("SiteID")["day"].nunique().rename("detection_days")

cal = cam.merge(detection_days, on="SiteID", how="left").fillna({"detection_days": 0})
cal["detection_days"] = cal["detection_days"].clip(upper=cal["trap_days"])

x_logit = to_logit(cal["suitability"].values)
X_cal = np.r_[x_logit, x_logit].reshape(-1, 1)
y_cal = np.r_[np.ones(len(cal)), np.zeros(len(cal))]
w_cal = np.r_[cal["detection_days"], cal["trap_days"] - cal["detection_days"]]
calibrator = LogisticRegression(C=np.inf).fit(X_cal, y_cal, sample_weight=w_cal)


score_range = (cal["suitability"].min(), cal["suitability"].max())


def daily_probability(score):
    score = np.clip(np.asarray(score), *score_range)
    return calibrator.predict_proba(to_logit(np.asarray(score)).reshape(-1, 1))[:, 1]


print(f"slope on logit(score): {calibrator.coef_[0, 0]:.2f} (> 0 means higher score -> more buffalo days)")
print(f"camera scores seen: {score_range[0]:.2f} to {score_range[1]:.2f} (grid scores outside this are capped)")

cal["predicted"] = daily_probability(cal["suitability"])
check = cal.groupby("suitability_group", observed=True).agg(
    observed=("detection_days", "sum"), trap_days=("trap_days", "sum"), predicted=("predicted", "mean"))
check["observed"] = check["observed"] / check["trap_days"]
print("\ndaily detection probability, observed vs calibrated:")
print((check[["observed", "predicted"]] * 100).round(2).astype(str) + " %")

grid["p_day"] = daily_probability(grid["suitability"])
print("\ngrid daily probability:")
print((grid["p_day"].describe(percentiles=[0.1, 0.5, 0.9])[["min", "10%", "50%", "90%", "max"]] * 100).round(2).astype(str) + " %")
```

## Focus area: re-run the model on a smaller box

The scores above are **relative to the whole Serengeti box**: the background was drawn from all visited cells, so the model learned "what makes buffalo spots different from the Serengeti as a whole". Zooming into the map only crops those scores.

Inside a smaller region the fair comparison is different: "what makes buffalo spots different from **this region**". So here the same pipeline is re-run with `FOCUS_BOUNDS` as the box:

1. Presences inside the focus box only (the thinned ones).
2. Background drawn only inside the focus box, with the same rules (`make_presence_background`).
3. Embeddings with the same Earth Engine function, an XGBoost model with the same settings, and block CV to check it.
4. A **finer grid** (`FOCUS_GRID_SPACING_M`, 1 km), scored by the local model **and** by the global model.

Comparing the two shows how much the local focus changes the ranking:
- A **high rank correlation** means the region's good spots were already the good spots in the global model.
- A **low one** means locally different habitat features matter.

**Pick the box from the heat map:** click a spot to see its lat/lon. A smaller box means fewer presences. Under `MIN_FOCUS_PRESENCES` (50) the local model has too little to learn from, so trust it less.

```python
FOCUS_BOUNDS = [34.6, -2.8, 35.3, -2.2]
FOCUS_GRID_SPACING_M = 1_000
MIN_FOCUS_PRESENCES = 50

f_min_lon, f_min_lat, f_max_lon, f_max_lat = FOCUS_BOUNDS


def in_focus(d):
    return d["decimalLongitude"].between(f_min_lon, f_max_lon) & d["decimalLatitude"].between(f_min_lat, f_max_lat)


focus_presences = thinned_df[in_focus(thinned_df)]
print("presences in focus box:", len(focus_presences), "of", len(thinned_df))
if len(focus_presences) < MIN_FOCUS_PRESENCES:
    print(f"WARNING: fewer than {MIN_FOCUS_PRESENCES} presences, the local model will be unreliable")

focus_points = make_presence_background(focus_presences, FOCUS_BOUNDS, visited_cells)

focus_info = get_range_year_reduced_area_embeddings(focus_points, "point_id", start_year=start_year, end_year=end_year)
focus_emb = pd.DataFrame([f["properties"] for year_info in focus_info for f in year_info["features"]])
focus_df = focus_points.merge(focus_emb, on="point_id", how="left").dropna(subset=band_cols).reset_index(drop=True)
print(focus_df["presence"].value_counts())
```

```python
from sklearn.base import clone

X_focus = focus_df[band_cols]
y_focus = focus_df["presence"]

fx, fy = to_utm.transform(focus_df["decimalLongitude"].values, focus_df["decimalLatitude"].values)
focus_cells = pd.Series((fx // cell_size_m).astype(int).astype(str)) + "_" + pd.Series((fy // cell_size_m).astype(int).astype(str))
n_splits = min(5, focus_cells[y_focus.values == 1].nunique())

if n_splits >= 3:
    focus_cv = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=42)
    auc_focus = cross_val_score(clone(xgb), X_focus, y_focus, cv=focus_cv, groups=focus_cells, scoring="roc_auc")
    print(f"local model, block CV AUC ({n_splits} folds): {auc_focus.mean():.3f} ± {auc_focus.std():.3f}")
else:
    print("presences sit in fewer than 3 cells: too few blocks for an honest CV score")

focus_model = clone(xgb).fit(X_focus, y_focus)
```

```python
offsets = np.arange(FOCUS_GRID_SPACING_M / 2, cell_size_m, FOCUS_GRID_SPACING_M)
focus_grid = pd.DataFrame(
    [(cx * cell_size_m + ox, cy * cell_size_m + oy) for cx, cy in visited_cells for ox in offsets for oy in offsets],
    columns=["x", "y"],
)
focus_grid["decimalLongitude"], focus_grid["decimalLatitude"] = to_lonlat.transform(focus_grid["x"].values, focus_grid["y"].values)
focus_grid = focus_grid[in_focus(focus_grid)].reset_index(drop=True)
focus_grid["point_id"] = "focus_" + focus_grid.index.astype(str)
focus_grid["year"] = MAP_YEAR
print("focus grid points:", len(focus_grid))

chunks = []
for start in range(0, len(focus_grid), GEE_CHUNK):
    info = get_range_year_reduced_area_embeddings(focus_grid.iloc[start:start + GEE_CHUNK], "point_id", start_year=MAP_YEAR, end_year=MAP_YEAR)
    chunks.append(pd.DataFrame([f["properties"] for f in info[0]["features"]]))
focus_grid = focus_grid.merge(pd.concat(chunks, ignore_index=True), on="point_id", how="left").dropna(subset=band_cols)

focus_grid["local"] = focus_model.predict_proba(focus_grid[band_cols])[:, 1]
focus_grid["global"] = final_model.predict_proba(focus_grid[band_cols])[:, 1]

rho_focus, _ = spearmanr(focus_grid["local"], focus_grid["global"])
print(f"rank correlation, local vs global: {rho_focus:.2f} (1 = same ranking)")
```
