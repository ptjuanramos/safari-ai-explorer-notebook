```python
import ee
import numpy as np

ee.Authenticate()
```

```python
ee.Initialize(project='project-0cf410a3-f35f-4911-9c5')
```

```python
col = ee.ImageCollection('GOOGLE/SATELLITE_EMBEDDING/V1/ANNUAL')
img = col.filterDate('2025-01-01', '2026-01-01').first()
```

```python
serengeti_square = ee.Geometry.Polygon([[
    [34.0, -1.6],
    [34.0, -3.6],
    [36.0, -3.6],
    [36.0, -1.6],
    [34.0, -1.6]
]])


points = ee.FeatureCollection([
    ee.Feature(serengeti_square, {'name': 'Serengheti'})
])
```

```python
print(type(points))
print(points)
```

```python
sampled = img.sampleRegions(collection=points, scale=10, geometries=False)

sampled.getInfo()
```

```python
year_imgs = (ee.ImageCollection('GOOGLE/SATELLITE_EMBEDDING/V1/ANNUAL')
.filterDate('2025-01-01', '2026-01-01')
.filterBounds(serengeti_square)
.mosaic())

year_imgs
```

```python
import geemap

Map = geemap.Map()
```

```python
visParams = {
    'min': -0.3, 
    'max': 0.3, 
    'bands': ['A01', 'A16', 'A09']
}

Map.addLayer(year_imgs.clip(serengeti_square), visParams, 'Embeddings Image')

Map.centerObject(serengeti_square, 12)

Map
```
