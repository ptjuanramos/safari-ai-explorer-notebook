```python
import osmnx as ox

TARANGIRE_BBOX = (-4.25, 35.78, -3.70, 36.08)

G = ox.graph_from_bbox(
    bbox=TARANGIRE_BBOX,
    network_type='drive',
    retain_all=True
)
print(f"Nodes: {G.number_of_nodes()}")
print(f"Edges: {G.number_of_edges()}")
```

```python
G = ox.graph_from_place(
    "Tarangire National Park, Tanzania",
    network_type='all',
    retain_all=True
)
```

```python
# Adding travel times and different set of speeds on each edge

G = ox.add_edge_speeds(G)
G = ox.add_edge_travel_times(G)
```

```python
print(f"Nodes: {G.number_of_nodes()}")
print(f"Edges: {G.number_of_edges()}")
```

```python
nodes, edges = ox.graph_to_gdfs(G)

print("=== EDGE ATTRIBUTES ===")
print(edges.columns.tolist())
```

```python
print("=== HIGHWAY TYPES ===")
print(edges['highway'].value_counts(dropna=False))
```

```python
print("=== SAMPLE EDGES ===")
print(edges.head(20))
```

```python
edges = edges.rename(columns={
    'highway': 'road_type',
    'length': 'length_m',
    'travel_time': 'travel_time_s'
})
```

```python
edges['travel_time_m'] = edges['travel_time_s'] / 60
```

- We currently have OSM data for each park location.
- We don't have the type of road.
- We kinda have some suspects on what they could be, however, it seems that we need to improve our implementation or model if we want to be more precise.
- For now, in order to reduce the number of edges, we remove the ones who are road_type = 'footway' and road_type = 'path'

```python
exclude_road_types = ['footway', 'path']

def is_vehicle_road(road_type) -> bool:
    return not any(r in exclude_road_types for r in road_type)
```

```python
edges_filtered = edges[edges['road_type'].apply(is_vehicle_road)]

print(f"NODES AFTER FILTER: {len(edges_filtered)}")
```

```python
#----------------------------------
# Focusing on core implementation
#----------------------------------

# This is fake of course
HIGHWAY_SURFACE_RISK = {
    'motorway':     0.0,
    'trunk':        0.0,
    'primary':      0.0,
    'secondary':    0.1,
    'tertiary':     0.2,
    'residential':  0.2,
    'service':      0.3,
    'unclassified': 0.4,
    'track':        0.8,
}
DEFAULT_SURFACE_RISK = 0.5

def compute_surface_risk(highway) -> float:
    if isinstance(highway, list):
        highway = highway[0]
    return HIGHWAY_SURFACE_RISK.get(str(highway), DEFAULT_SURFACE_RISK)

edges_filtered['surface_risk'] = edges_filtered['road_type'].apply(
    compute_surface_risk
)

print(edges_filtered[['road_type', 'length_m', 'travel_time_m', 'surface_risk']].head(10))
print(f"\nSurface risk distribution:")
print(edges_filtered['surface_risk'].value_counts())
```

```python
RISK_WEIGHT = 60.0  # minutes penalty for risk = 1.0

def compute_edge_cost(travel_time_minutes: float, surface_risk: float) -> float:
    return travel_time_minutes + (RISK_WEIGHT * surface_risk)

edges_filtered['cost'] = edges_filtered.apply(
    lambda row: compute_edge_cost(row['travel_time_m'], row['surface_risk']),
    axis=1
)

print(edges_filtered[['road_type', 'travel_time_m', 'surface_risk', 'cost']].head(10))
print(f"\nMean cost (min): {edges_filtered['cost'].mean():.1f}")
print(f"Max cost (min):  {edges_filtered['cost'].max():.1f}")
print(f"Min cost (min):  {edges_filtered['cost'].min():.1f}")
```

```python
m = folium.Map(location=[center_lat, center_lon], zoom_start=10)

COST_COLORS = {
    (0, 20):   '#2ecc71',   # low cost
    (20, 40):  '#f39c12',   # medium cost
    (40, 80):  '#e74c3c',   # high cost
    (80, 999): '#7b241c',   # very high cost
}

def get_cost_color(cost: float) -> str:
    for (low, high), color in COST_COLORS.items():
        if low <= cost < high:
            return color
    return '#7b241c'

for _, row in edges_filtered.iterrows():
    coords = [(lat, lon) for lon, lat in row['geometry'].coords]
    color  = get_cost_color(row['cost'])

    folium.PolyLine(
        locations=coords,
        color=color,
        weight=3,
        opacity=0.8,
        popup=(
            f"highway: {row['road_type']}<br>"
            f"travel_time: {round(row['travel_time_m'], 1)} min<br>"
            f"surface_risk: {row['surface_risk']}<br>"
            f"cost: {round(row['cost'], 1)} min"
        )
    ).add_to(m)

sample_nodes = nodes.sample(50, random_state=42)

for idx, row in sample_nodes.iterrows():
    folium.CircleMarker(
        location=[row['y'], row['x']],
        radius=5,
        color='steelblue',
        fill=True,
        fill_opacity=0.7,
        popup=f"Node ID: {idx}"
    ).add_to(m)

m
```
