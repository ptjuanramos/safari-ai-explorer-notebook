```python
from antecedent_precipitation_index import AntecedentPrecipitationIndex
import numpy as np
import matplotlib.pyplot as plt
```

```python
api = AntecedentPrecipitationIndex()

t_clt = t_avg = 20.0
k = api._recession_coefficient(t_clt, t_avg)

days = np.arange(0, 31)
weights = k ** days

plt.plot(days, weights, marker="o", markersize=3)
plt.xlabel("days ago (d)")
plt.ylabel("decay weight k^d")
plt.title(f"Decay weight vs. days ago (k = {k:.3f})")
plt.axhline(0.1, color="gray", linestyle="--", linewidth=1, label="10% of original weight")
plt.legend()
plt.show()
```

```python
event_rain = 50.0
t_clt = t_avg = 20.0

contributions = [
    api._compute_single(d, event_rain, t_clt, t_avg)
    for d in range(0, 22)
]

plt.bar(range(len(contributions)), contributions, color="#2f6fb0")
plt.xlabel("days since the 50mm event")
plt.ylabel("contribution to rain_mm")
plt.title("How a single rain event fades out of the index")
plt.show()
```
